```python
# Vstack - https://numpy.org/doc/stable/reference/generated/numpy.vstack.html
matrix1 = np.array([[1, 2, 3], [1, 2, 3]])
matrix2 = np.array([[5, 6, 7], [5, 6, 7]])

print(matrix1.shape)
print(matrix2.shape)
print('\n')

vstacked = np.vstack([matrix1, matrix2])
print(vstacked) # 2D arrays must have the same 2nd dimension (number of columns)
print(vstacked.shape)
```

```python
# Hstack - https://numpy.org/doc/stable/reference/generated/numpy.hstack.html
matrix1 = np.array([[1, 2, 3], 
                    [1, 2, 3]])

matrix2 = np.array([[5, 6, 7], 
                    [5, 6, 7]])

print(matrix1.shape)
print(matrix2.shape)
print('\n')

hstacked = np.hstack([matrix1, matrix2]) # 2D arrays must have the same 1st dimension (number of rows)
print(hstacked) 
print(hstacked.shape)
```

```python
# Concatenate - https://numpy.org/doc/stable/reference/generated/numpy.concatenate.html
matrix1 = np.array([[[1],[1], [1]], [[1], [1], [1]]])
matrix2 = np.array([[[2], [2], [2], [2]], [[2], [2], [2], [2]]])
matrix3 = np.array([[[3], [3], [3]], [[3], [3], [3]]])

print(matrix1.shape)
print(matrix2.shape)
print(matrix3.shape)
print('\n')

# Arrays must have the same dimensions except for the concatenation axis
concatenated = np.concatenate([matrix1, matrix2, matrix3], axis=1)
print(concatenated)
print(concatenated.shape)
```

#### Broadcasting (performing vectorized operations with arrays of different shapes)

```python
# Example of a binary operation with arrays of different shapes

arr1 = np.array([8,4,2])
print(arr1.shape)
arr2 = np.array([2])
print(arr2.shape)
print("\n")

print(arr1+arr2)
```

```python
# Example of a binary operation with arrays of different shapes and ndim

arr1 = np.array([[8,4,2],[16,8,4]])
print(arr1.shape)
arr2 = np.array([1,2,3])
print(arr2.shape)
print("\n")

print(np.array([[8,4,2],[16,8,4]]) + np.array([1,2,3]))
print("\n")
```

**High-Level Broadcasting Pseudocode with binary operators**

**(1)** Align dimensions

If arrays have different numbers of dimensions, left-pad the smaller shape with 1s so that the two arrays have the same number of axes.

**(2)** Check compatibility

For each dimension, starting from the right and moving towards the left, the sizes must be equal or at least one of them must be 1. If any dimensions fails this test, broadcasting is not possible.

**(3)** Broadcast

For every dimension with non-matching sizes, create copies of the array with size 1 until matching the largest size.
